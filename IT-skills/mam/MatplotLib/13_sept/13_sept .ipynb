```python
Question 2: Plot a line chart using matplotlib
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

months = df['Month'].tolist()
sales = df['Sales'].tolist()

plt.plot(months,sales , marker='o' , color="red")
plt.title("Monthly Sales Data")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True , alpha = 0.5)
plt.tight_layout()
plt.show()
```

Q3: bar chart instead of a line chart

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

months = df['Month'].tolist()
sales = df['Sales'].tolist()

plt.bar(months,sales , color="green")
plt.title("Monthly Sales Data - bar chart")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

Q4 : Modify the bar chart to highlight the month with the highest sales in red, and the rest in blue.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

months = df['Month'].tolist()
sales = df['Sales'].tolist()
colors = ['red' if sale == max(sales) else 'blue' for sale in sales]

plt.bar(months,sales ,color=colors)
plt.title("Monthly Sales Data - Hightest sale")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

Question:5 Save the sales bar chart to a PNG file called sales_chart.png instead of displaying it.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

months = df['Month'].tolist()
sales = df['Sales'].tolist()
c='pink'
plt.bar(months,sales ,color=c)
plt.title("Monthly Sales Data - Hightest sale")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.savefig('sales_chart.png')
print("Chart saved")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

months = df['Month'].tolist()
sales = df['Sales'].tolist()

plt.pie(sales ,labels=months  , autopct='%0.1f%%')
plt.title("Sales Contribution Monthly")
plt.axis('equal')
plt.tight_layout()
plt.show()
```
