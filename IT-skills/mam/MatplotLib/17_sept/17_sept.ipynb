10.1 using lambda . Add a New Column for Pass/Fail (Pass if Marks ≥ 80)
 Question:Add a column named Result that contains Pass or Fail.

```python
df['Result'] = df['Marks'].apply(lambda x: 'Pass' if x >= 80 else 'Fail')
print(df)
```
