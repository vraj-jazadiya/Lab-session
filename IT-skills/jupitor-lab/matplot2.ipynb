```python
import matplotlib.pyplot as plt
import numpy as np

#data
a = np.arange(5)
b = [1,12,3,14,5]
c = [6,7,8,9,10]

fig = plt.figure()
ax = plt.subplot()

ax.plot(a,b , 'y--' , label="Frequency")
ax.plot(a,c , 'k--' , label="Period")
ax.legend(loc= "upper left" , fontsize='small')

plt.title("Frequency of Signal")

plt.show()

```
