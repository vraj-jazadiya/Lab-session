```python
import matplotlib.pyplot as plt
import pandas as pd

dataframe = pd.DataFrame(
    {
        'name': ['Arjun', 'Sneha', 'Ravi', 'Priya', 'Karan', 'Anita', 'Vikram'],
        'age': [25, 22, 28, 24, 30, 27, 26]
    }
)
xpts = dataframe["name"]
ypts = dataframe["age"]

plt.title("Data of Student") # postion left or right using - loc=""

plt.xlabel("student's Name")
plt.ylabel("student's Age")

plt.plot(xpts, ypts, color="blue", ls="-.", linewidth=2.5)

plt.grid(axis="x", color="green", ls="--", linewidth=0.5)

plt.show()
```

```python
# adding legend in gharph
import matplotlib.pyplot as plt
import numpy as np

#data
a = np.arange(5)
b = [1,12,3,14,5]
c = [6,7,8,9,10]

fig = plt.figure()
ax = plt.subplot()

ax.plot(a,b,'k--',label='Frequency')
ax.plot(a,c,'r-.',label='Periods')


ax.legend()

plt.title("Frequency of Signal")

plt.show()


```

```python
import matplotlib.pyplot as plt
import numpy as np

#data
a = np.arange(5)
b = [1,12,3,14,5]
c = [6,7,8,9,10]

fig = plt.figure()
ax = plt.subplot()

ax.plot(a,b , 'y--' , label="Frequency")
ax.plot(a,c , 'k--' , label="Period")
ax.legend(loc= "upper center")

plt.title("Frequency of Signal")

plt.show()

```
