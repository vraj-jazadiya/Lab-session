```python
#Q1

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10,100)
y1 = x**2
y2 = x**3
y3 = np.sqrt(x)

plt.plot(x,y1 , 'r--', label='x^2')
plt.plot(x,y2 , 'g--', label='x^3')
plt.plot(x,y3 , 'b--', label='sqet(x)')

plt.title("Multiple Line plot")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()

```

```python
#Q2

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi ,100)
y1 = np.sin(x)
y2 = np.cos(x)

fig , (ax1 , ax2) = plt.subplots(2,1 ,sharex=True)

ax1.plot(x,y1 ,'r')
ax1.set_title("Sine fuction")

ax2.plot(x,y2 ,'r')
ax2.set_title("Cosine fuction")

plt.xlabel('x')
plt.show()
```

```python
#Q3

xsubject = ['math','sci','eng']
avg_score = [85,90,78]
error = [5,7,4]

plt.bar(xsubject , avg_score , yerr=error , width=0.8 , capsize=5)
plt.show()
```

```python
x = np.random.uniform(0,10,100)
y = np.random.uniform(0,10,100)
dist = np.sqrt(x**2  + y**2)

plt.scatter(x,y,c=dist , cmap="gist_heat")
plt.colorbar(label="distance from orgin")
```

```python
#q5

data1= np.random.normal(0,1,1000)
data2= np.random.normal(2,1.5,1000)

plt.hist(data1,bins=30 , color="r" , alpha=0.8 , label="data1")
plt.hist(data2,bins=30 , color="k" , alpha=0.8 , label="data2")
plt.legend()
plt.title("Overlapping histograms")
plt.show()
```

```python
#B-2

x = np.random.rand(100)
y = np.random.rand(100)
dist = np.sqrt(x**2  + y**2)

plt.scatter(x,y,s=dist*500,c=dist , cmap="plasma", alpha=0.6)
plt.colorbar(label="distance from orgin")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10,100)
y1 = x
y2 = x**2
y3 = x**3

fig ,axs = plt.subplots(3,1,sharex=True)

axs[0].plot(x,y1 , 'r--', label='x^2')
axs[1].plot(x,y2 , 'g--', label='x^3')
axs[2].plot(x,y3 , 'b--', label='sqet(x)')

plt.xlabel('x')
fig.legend()

axs[0].grid()
plt.tight_layout()
plt.show()
```
