# Text in matplotlib

```python
    import matplotlib.pyplot as plt
    import numpy as np

    # Example data
    x = np.random.rand(100)
    y = np.random.rand(100)
    c = np.random.rand(100) # Data to be mapped to color

    plt.scatter(x, y, c=c, cmap='viridis') # Using the 'viridis' colormap
    plt.colorbar(label='Data Value') # Add a colorbar to show the mapping
    plt.title('Scatter Plot with Viridis Colormap')
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(5)
y = [30,80,60,40,50]

plt.ylim(0,100)

lineplot = plt.plot(x,y , marker="o" , label="vraj")
plt.xlabel("X-ray")
plt.ylabel("Y-ray")
plt.legend()
plt.title("Just a random line plot")

plt.text(2,85,"point")
plt.annotate("check",xy=(1,80) ,xytext=(2,80) , arrowprops=dict(facecolor='b',shrink=0.05) , bbox=dict(facecolor='red', alpha=0.5))

plt.show()
```
