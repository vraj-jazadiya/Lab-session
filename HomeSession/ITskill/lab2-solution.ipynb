7) Write a Pandas program to join the two dataframes with matching records from both sides where available.

```python
df5 = pd.DataFrame(df4.head())
df5= df5.reset_index()
df5 = df5.rename(columns={"index":"name"})
print(df5)

df6 = pd.DataFrame(df4.tail())
df6= df6.reset_index()
df6 = df6.rename(columns={"index":"name"})
print(df6)
```

```python
df7 = pd.merge(df5,df6 , on=["name","score","qualify","attempts","Country"] , how="outer")
print(df7)
```
