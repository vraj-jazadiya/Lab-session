```python
import matplotlib.pyplot as plt
import numpy as np

x = ['A','B','C','D','E','F']
y = [8,2,3,4,5,6]
z = [7,4,8,3,2,4]

width=0.3

p = np.arange(len(x))
p1 = [j + width for j in p]

plt.xlabel("X-Name A-Z", fontsize=12)
plt.ylabel("Y-Number for A-Z", fontsize=12)
plt.title("Relation of A-Z", fontsize=20)

plt.bar( p,y  , width , color='r' , label="vraj")
plt.bar( p1,z , width , color='y' , label="jarv")

plt.xticks(p+width/2,x,rotation=-20)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = ['A','B','C','D','E','F']
y = [8,2,3,4,5,6]
z = [7,4,8,3,2,4]

width=0.3

p = np.arange(len(x))
p1 = [j + width for j in p]

plt.xlabel("X-Name A-Z", fontsize=12)
plt.ylabel("Y-Number for A-Z", fontsize=12)
plt.title("Relation of A-Z", fontsize=20)

plt.barh( p,y  , width , color='r' , label="vraj")
plt.barh( p1,z , width , color='y' , label="jarv")

plt.xticks(p+width/2,x,rotation=-20)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

a = np.arange(10)
b = np.array([10,44,54,58,54,48,16,65,54,49])

colors = ['r','y','b','k','b','r','m','r','k','y']
sizes = [45,554,44,47,87,55,45,54,8,48]

plt.scatter(a,b , s=sizes, marker='o' , c=colors)
plt.title("Scatter Plot" , fontsize=15)
plt.xlabel("x", fontsize=12 , loc='right')
plt.ylabel("y", fontsize=12 , loc='top')

plt.grid()
plt.show()
```
